```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "Age group": ["Children <18", "Adults 18-64", "Adults 65+"],
    "Total cases": [294, 1397, 132],
    "Severe cases": [48, 177, 26]
})

display(df)
```

```python
df["Severe (%)"] = 100 * df["Severe cases"] / df["Total cases"]

display(df.round(2))
```

```python
df.plot(x="Age group", y="Severe (%)", kind="bar", legend=False)

plt.title("Malaria Severity by Age Group — USA, 2018")
plt.ylabel("Severe cases (%)")
plt.xlabel("Age group")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
